```python
from roman_mast import list_data
from roman_fits import stream_materialized, to_imviz, download_catalogs
from IPython.display import display, HTML
display(HTML(
    "<style>.cell-output-ipywidget-background,"
    ".jp-OutputArea-output{max-width:100% !important;}</style>"
))


res = list_data(program=161, sca_only=True)
exp = res.select(1)

dm_dict = stream_materialized(exp, res.missions, scas=[3], max_workers=20)
cat_paths = download_catalogs(exp, res.missions, scas=[3], out_dir='./catalog')

app = to_imviz(dm_dict, exp, dq_overlay=True, catalog_paths=cat_paths, wcs_link=False)
app.show(height=900)  
```
