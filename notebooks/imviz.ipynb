```python
from roman_mast import list_data
from roman_fits import stream_materialized, to_imviz, download_catalogs
from IPython.display import display, HTML
display(HTML(
    "<style>.cell-output-ipywidget-background,"
    ".jp-OutputArea-output{max-width:100% !important;}</style>"
))


res = list_data(program=161, sca_only=True)
exp = res.select(1)

dm_dict = stream_materialized(exp, res.missions, scas=[3], max_workers=20)
cat_paths = download_catalogs(exp, res.missions, scas=[3], out_dir='./catalog')

app = to_imviz(dm_dict, exp, dq_overlay=True, catalog_paths=cat_paths, wcs_link=False)
app.show(height=900)  
viewer_id = 'Image'
for label in [d.label for d in app._app.data_collection]:
    # Skip catalog marker data — those go through add_markers, not the image viewer.
    if label.startswith('cat_'):
        continue
    try:
        app._app.add_data_to_viewer(viewer_id, label)
    except Exception as e:
        print(f"  skipped {label}: {type(e).__name__}: {e}")

# Force a redraw
v = app.viewers[viewer_id]
v.center_on((2044, 2044))
```

```python
# Is the data actually loaded?
print("data collection:", [d.label for d in app._app.data_collection])

# Is there a viewer?
print("viewers:", list(app.viewers.keys()) if hasattr(app, 'viewers') else "none")
```
